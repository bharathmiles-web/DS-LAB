```python
class Node:

    def __init__(self, data):
        self.data = data
        self.next = None


class Queue:

    def __init__(self):
        self.front = self.rear = None

    def isEmpty(self):
        return self.front == None

    def EnQueue(self, item):

        temp = Node(item)

        # If queue is empty
        if self.rear == None:
            self.front = self.rear = temp
            return

        # Add new node at the rear
        self.rear.next = temp
        self.rear = temp

    def DeQueue(self):

        # If queue is empty
        if self.isEmpty():
            return

        temp = self.front

        # Move front to next node
        self.front = temp.next

        # If queue becomes empty
        if self.front == None:
            self.rear = None

        return str(temp.data)


if __name__ == '__main__':

    q = Queue()

    q.EnQueue(10)
    q.EnQueue(20)

    q.DeQueue()
    q.DeQueue()

    q.EnQueue(30)
    q.EnQueue(40)
    q.EnQueue(50)

    print("Dequeued item is " + q.DeQueue())
```
